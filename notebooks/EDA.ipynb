Imprting necessaary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

path setup

```python
csv_path = "../Practice/EchoNet-Dynamic/FileList.csv"
```

loading dataset

```python
df = pd.read_csv(csv_path)
```

data structure

```python
print(df.info())
print(df[['EF', 'EDV', 'ESV']].describe())
```

Visualizations

```python
plt.figure(figsize=(12, 6))

sns.histplot(data=df, x='EF', color='teal', bins=30, kde=True) #bar plot
sns.kdeplot(data=df, x='EF', color='orange', linewidth=2, label='KDE')  #curve plot over bars

plt.axvline(x=50, color='red', linestyle='--', label='EF = 50%') #clinical boundry for normal EF

plt.title('Distribution of Ejection Fraction (EF)')
plt.xlabel('Ejection Fraction (EF)%')
plt.ylabel('Number of patients')
plt.legend()
plt.show()

#checking data split
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Split', palette='Set2')
plt.title('Patient Allocation per ML Split')
plt.show()
```
